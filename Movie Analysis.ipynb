# **Movie Analysis Project**

## **Dataset Overview**
The dataset is taken from *Kaggle.com* and contains ***1000 top-rated IMDb movies*** with details like *rating*, *genre*, and *cast*.

## **Objective**
The Objective of Cleaning and Analysis of this Dataset is to find the watch Trends, Relation and Impact of Genre, Director, Star Cast on the Revenue and Performance of the Movie.

## Key Features
- Title
- Year of release
- Star Cast
- IMDB rating
- Gross revenue

# **Libraries Used and Loading Dataset**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb

Og_ds = pd.read_csv('C:/Users/Dell/Downloads/IMDb Top 1000.csv')
Og_ds.info()
Og_ds.describe()
```

# **Reading Datset**

```python
Og_ds.head(5)
```

# **Data Cleaning**
## Slicing of Rows from the Dataset

```python
Cleaned_ds1 = Og_ds.iloc[:140]
Cleaned_ds2= Og_ds.iloc[-16:]
Cleaned_ds= pd.concat([Cleaned_ds1, Cleaned_ds2],axis=0)
Cleaned_ds.info()
Cleaned_ds.head(5)
```

# **Data Manupilation**

## Identifying Null Values in Individual Columns

```python
Cleaned_ds.isna().sum()
```

# Removing Unnecessary Columns

```python
Cleaned_ds.drop(['Poster_Link','Certificate','Runtime','Overview','Meta_score','Star3','Star4'],axis=1, inplace=True)
Cleaned_ds.columns.to_list()
```

## Checking for Duplicate Values

```python
Cleaned_ds.duplicated().sum()
```

## Previewing Duplicate Values

```python
Cleaned_ds[Cleaned_ds.duplicated()] # No duplicates
```

## Checking for Rows with Null Values

```python
Cleaned_ds[Cleaned_ds.isnull().any(axis=1)]
```

## Filling Null Values in ***Gross***

Creating a Backup In Case of Errors

```python
#Cleaned_ds['Gross_backup'] = Cleaned_ds['Gross'].copy()
Cleaned_ds['Gross'].head(5)
```

Checking DataType of Gross

```python
Cleaned_ds['Gross'].dtypes
```

Checking Unique Value for Dtype ***Object***

```python
Cleaned_ds['Gross'].unique()
```

**Removing Commas and Spaces Created by them**

```python
Cleaned_ds['Gross'] = Cleaned_ds['Gross'].str.replace('[,]', '', regex=True).str.strip()
Cleaned_ds['Gross'] = Cleaned_ds['Gross'].replace('', pd.NA)
```

Checking Unique Values again after Replacment

```python
Cleaned_ds['Gross'].unique()
```

**Converting Gross from Object type to Numeric type**

```python
Cleaned_ds['Gross'] = pd.to_numeric(Cleaned_ds['Gross'], errors='coerce')
Cleaned_ds['Gross'].head(5)
```

**Checking Dtype again for Verification**

```python
Cleaned_ds['Gross'].dtypes
```

Checking Null Values before Filling

```python
Cleaned_ds['Gross'].isna().sum()
```

**Filling NaN with Median in Gross**

```python
Cleaned_ds['Gross'] = Cleaned_ds['Gross'].fillna(Cleaned_ds['Gross'].median())
Cleaned_ds['Gross'].isna().sum()
```

Checking Null Value after Filling

```python
Cleaned_ds['Gross'].isnull().sum()
```

```python
Cleaned_ds['Gross'].head(5)
```

Rows with Values of Gross Changed to Median

```python
median_value = Cleaned_ds['Gross'].median()
Cleaned_ds[Cleaned_ds['Gross'] == median_value][['Gross']]
```

Checking Movie with Same Gross

```python
Cleaned_ds[Cleaned_ds['Gross'] == Cleaned_ds['Gross'].median()][['Series_Title', 'Gross']]
```

**Renaming Columns**

```python
Cleaned_ds = Cleaned_ds.rename(columns={
    'Series_Title': 'Movie Title',
    'Released_Year': 'Release Year',
    'Genre': 'Genres',
    'IMDB_Rating': 'IMDb Rating',
    'Star1': 'Actor 1',
    'Star2': 'Actor 2',
    'No_of_Votes': 'User Votes',
    'Gross': 'Box Office Collections',
})
Cleaned_ds.columns.to_list()
```

```python
Cleaned_ds.to_csv('Cleaned Dataset.csv',index=False)
```
